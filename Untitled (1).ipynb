```python

# ============================================================
# PROJECT: ANALYSING PHARMACEUTICAL SALES DATA
# Libraries: Pandas and Matplotlib
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# ------------------------------------------------------------
# STEP 1: READ ALL FOUR CSV FILES
# ------------------------------------------------------------

data_folder = Path(
    r"D:\Aadesh\Analysing-pharmaceutical-sales-data\data"
)

daily_path = data_folder / "salesdaily.csv"
hourly_path = data_folder / "saleshourly.csv"
monthly_path = data_folder / "salesmonthly.csv"
weekly_path = data_folder / "salesweekly.csv"

# Check that all files exist
file_paths = {
    "Daily": daily_path,
    "Hourly": hourly_path,
    "Monthly": monthly_path,
    "Weekly": weekly_path
}

for name, path in file_paths.items():
    if not path.exists():
        raise FileNotFoundError(
            f"{name} dataset not found: {path}"
        )

# Read all datasets
df = pd.read_csv(daily_path)
df_hourly = pd.read_csv(hourly_path)
df_monthly = pd.read_csv(monthly_path)
df_weekly = pd.read_csv(weekly_path)

# Clean column names
for dataset in [df, df_hourly, df_monthly, df_weekly]:
    dataset.columns = dataset.columns.astype(str).str.strip()

# Display dataset information
print("=" * 65)
print("ALL DATASETS LOADED SUCCESSFULLY")
print("=" * 65)

for name, dataset in [
    ("DAILY", df),
    ("HOURLY", df_hourly),
    ("MONTHLY", df_monthly),
    ("WEEKLY", df_weekly)
]:
    print(f"\n{name} DATASET")
    print("Shape:", dataset.shape)
    print("Columns:", dataset.columns.tolist())
    display(dataset.head())

# ------------------------------------------------------------
# STEP 2: PREPARE DAILY DATA
# ------------------------------------------------------------

df["datum"] = pd.to_datetime(df["datum"], errors="coerce")
df = df.dropna(subset=["datum"]).sort_values("datum").copy()

drug_categories = [
    "M01AB", "M01AE", "N02BA", "N02BE",
    "N05B", "N05C", "R03", "R06"
]

missing_categories = [
    c for c in drug_categories if c not in df.columns
]

if missing_categories:
    raise ValueError(
        f"Missing expected ATC columns: {missing_categories}"
    )

for category in drug_categories:
    df[category] = pd.to_numeric(
        df[category], errors="coerce"
    ).fillna(0)

df["Year"] = df["datum"].dt.year
df["Month"] = df["datum"].dt.month

print("\nDaily dataset shape after cleaning:", df.shape)
print("Date range:", df["datum"].min().date(),
      "to", df["datum"].max().date())


# ------------------------------------------------------------
# QUESTION 1: TOTAL SALES FOR EACH ATC CATEGORY
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("QUESTION 1: TOTAL SALES FOR EACH DRUG CATEGORY")
print("=" * 65)

total_sales = df[drug_categories].sum().sort_values(
    ascending=False
)

display(total_sales.rename("Total Sales Quantity").to_frame())

plt.figure(figsize=(10, 5))
total_sales.plot(kind="bar", edgecolor="black")
plt.title("Total Sales by ATC Drug Category")
plt.xlabel("ATC Drug Category")
plt.ylabel("Total Sales Quantity")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# QUESTION 2: INDIVIDUAL DRUG BRANDS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("QUESTION 2: INDIVIDUAL DRUG BRANDS WITH HIGHEST SALES")
print("=" * 65)

print(
    "The four datasets contain ATC drug category sales, "
    "not individual drug brand names."
)
print("\nHighest-selling ATC categories:")
display(total_sales.head(3).rename("Total Sales").to_frame())


# ------------------------------------------------------------
# QUESTION 3: TOP THREE CATEGORIES IN SELECTED MONTHS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("QUESTION 3: TOP THREE CATEGORIES IN SELECTED MONTHS")
print("=" * 65)


def top_three_monthly_sales(year, month):
    selected = df[
        (df["Year"] == year) &
        (df["Month"] == month)
    ]

    month_name = pd.Timestamp(
        year, month, 1
    ).strftime("%B %Y")

    print(f"\nTop 3 categories for {month_name}:")

    if selected.empty:
        print("No data available for this period.")
        return pd.Series(dtype=float)

    result = (
        selected[drug_categories]
        .sum()
        .sort_values(ascending=False)
        .head(3)
    )

    display(result.rename("Total Sales").to_frame())
    return result


jan_2015 = top_three_monthly_sales(2015, 1)
jul_2016 = top_three_monthly_sales(2016, 7)
sep_2017 = top_three_monthly_sales(2017, 9)

monthly_results = {
    "January 2015": jan_2015,
    "July 2016": jul_2016,
    "September 2017": sep_2017
}

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for ax, (period, result) in zip(
    axes, monthly_results.items()
):
    if not result.empty:
        result.plot(kind="bar", ax=ax, edgecolor="black")
        ax.set_title(period)
        ax.set_xlabel("ATC Category")
        ax.set_ylabel("Total Sales")
        ax.tick_params(axis="x", rotation=45)
    else:
        ax.set_title(period)
        ax.text(0.5, 0.5, "No data available",
                ha="center", va="center")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# QUESTION 4: HIGHEST-SELLING CATEGORY IN 2017
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("QUESTION 4: HIGHEST SALES IN 2017")
print("=" * 65)

data_2017 = df[df["Year"] == 2017]

if data_2017.empty:
    print("No data available for 2017.")
else:
    sales_2017 = (
        data_2017[drug_categories]
        .sum()
        .sort_values(ascending=False)
    )

    display(sales_2017.rename("2017 Total Sales").to_frame())

    print("Highest-selling category:", sales_2017.idxmax())
    print("Total sales:", round(sales_2017.max(), 2))

    sales_2017.plot(
        kind="bar", figsize=(10, 5), edgecolor="black"
    )
    plt.title("Total Drug Category Sales in 2017")
    plt.xlabel("ATC Drug Category")
    plt.ylabel("Total Sales Quantity")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# QUESTION 5: HIGHEST AVERAGE DAILY SALES
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("QUESTION 5: HIGHEST AVERAGE DAILY SALES")
print("=" * 65)

average_daily_sales = (
    df[drug_categories]
    .mean()
    .sort_values(ascending=False)
)

display(
    average_daily_sales
    .round(2)
    .rename("Average Daily Sales")
    .to_frame()
)

print("Highest average daily sales category:",
      average_daily_sales.idxmax())
print("Average daily sales:",
      round(average_daily_sales.max(), 2))

average_daily_sales.plot(
    kind="bar", figsize=(10, 5), edgecolor="black"
)
plt.title("Average Daily Sales by ATC Category")
plt.xlabel("ATC Drug Category")
plt.ylabel("Average Daily Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# QUESTION 6: RESPIRATORY DRUG SALES (R03) BY MONTH
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("QUESTION 6: MONTHLY RESPIRATORY DRUG SALES (R03)")
print("=" * 65)

month_names = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

r03_average = (
    df.groupby("Month")["R03"]
    .mean()
    .reindex(range(1, 13))
)

r03_total = (
    df.groupby("Month")["R03"]
    .sum()
    .reindex(range(1, 13))
)

r03_summary = pd.DataFrame({
    "Month": month_names,
    "Average Daily Sales": r03_average.values,
    "Total Sales Across All Years": r03_total.values
})

display(r03_summary.round(2))

highest_month = r03_average.idxmax()
lowest_month = r03_average.idxmin()

print("Highest average daily R03 sales:",
      month_names[highest_month - 1])
print("Lowest average daily R03 sales:",
      month_names[lowest_month - 1])

plt.figure(figsize=(11, 5))
plt.plot(
    month_names,
    r03_average.values,
    marker="o",
    linewidth=2
)
plt.title("Average Daily Respiratory Drug (R03) Sales by Month")
plt.xlabel("Month")
plt.ylabel("Average Daily Sales")
plt.xticks(rotation=45)
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

plt.figure(figsize=(11, 5))
plt.bar(month_names, r03_total.values, edgecolor="black")
plt.title("Total R03 Sales by Calendar Month Across All Years")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# FINAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("FINAL SUMMARY")
print("=" * 65)

print("\n1. Total sales by category:")
display(total_sales.rename("Total Sales").to_frame())

print("\n2. Individual brands:")
print("Cannot be identified from these datasets.")

print("\n3. Top categories:")
print("January 2015:", jan_2015.to_dict())
print("July 2016:", jul_2016.to_dict())
print("September 2017:", sep_2017.to_dict())

if not data_2017.empty:
    print("\n4. Highest-selling category in 2017:",
          sales_2017.idxmax())

print("\n5. Highest average daily sales category:",
      average_daily_sales.idxmax())

print("\n6. Highest average R03 sales month:",
      month_names[highest_month - 1])

print("\nAll analysis completed successfully!")
```
